```python
# Tính giá trị trung bình của cột Newspaper
newspaper_mean = np.mean(data[:, 2])

# Tính tổng các giá trị Sales với điều kiện Newspaper > giá trị trung bình
sales_sum_cond = np.sum(data[data[:, 2] > newspaper_mean][:, 3])
sales_sum_cond

```

```python
# Tính giá trị trung bình của cột Sales
sales_mean = np.mean(data[:, 3])

# Tạo mảng scores
scores = np.where(data[:, 3] < sales_mean,'Bad', np.where(data[:, 3] > sales_mean, 'Good', 'Average'))

# In kết quả scores[7:10]
scores[7:10]
```

```python
# Tìm giá trị gần nhất với giá trị trung bình của cột Sales
sales_mean = np.mean(data[:, 3])
closest_to_mean = data[:, 3][np.abs(data[:, 3] - sales_mean).argmin()]

# Tạo mảng scores
scores_closest = np.where(data[:, 3] > closest_to_mean, 'Good', np.where(data[:, 3] < closest_to_mean, 'Bad', 'Average'))

# In kết quả scores[7:10]
scores_closest[7:10]
```
